```python

import pandas as pd
import numpy as np

# Make the results reproducible
np.random.seed(42)

# Create sample sales data
n = 500

products = ["Laptop", "Mobile Phone", "Headphones", "Tablet", "Keyboard"]
categories = ["Electronics", "Electronics", "Accessories", "Electronics", "Accessories"]
regions = ["North", "South", "East", "West"]
customers = [f"Customer_{i:03d}" for i in range(1, 101)]

product_prices = {
    "Laptop": 50000,
    "Mobile Phone": 25000,
    "Headphones": 2000,
    "Tablet": 18000,
    "Keyboard": 1500
}

df = pd.DataFrame({
    "Order_ID": range(1001, 1001 + n),
    "Order_Date": pd.to_datetime(
        np.random.choice(
            pd.date_range("2025-01-01", "2025-12-31"),
            n
        )
    ),
    "Customer": np.random.choice(customers, n),
    "Product": np.random.choice(products, n),
    "Region": np.random.choice(regions, n),
    "Quantity": np.random.randint(1, 6, n)
})

# Add category, sales and profit
df["Category"] = df["Product"].map(dict(zip(products, categories)))
df["Sales"] = df["Product"].map(product_prices) * df["Quantity"]
df["Profit"] = (df["Sales"] * np.random.uniform(0.08, 0.25, n)).round(2)

# Save a copy of the dataset
df.to_csv("customer_sales_data.csv", index=False)

# Display the first five records
df.head()

```

```python

# Calculate key business metrics

total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order_ID"].nunique()
total_customers = df["Customer"].nunique()
total_quantity = df["Quantity"].sum()
average_order_value = total_sales / total_orders

print("CUSTOMER SALES DASHBOARD")
print("-" * 35)
print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Total Profit: ₹{total_profit:,.2f}")
print(f"Total Orders: {total_orders}")
print(f"Unique Customers: {total_customers}")
print(f"Units Sold: {total_quantity}")
print(f"Average Order Value: ₹{average_order_value:,.2f}")
```

```python

import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# Group sales by month
monthly_sales = (
    df.set_index("Order_Date")["Sales"]
    .resample("MS")
    .sum()
)

# Create the chart
plt.figure(figsize=(12, 5))
plt.plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    linewidth=2
)

plt.title("Monthly Sales Trend - 2025", fontsize=16)
plt.xlabel("Month")
plt.ylabel("Total Sales (₹)")
plt.gca().yaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, pos: f"₹{x:,.0f}")
)
plt.xticks(monthly_sales.index, monthly_sales.index.strftime("%b"), rotation=0)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Calculate total sales for each product
product_sales = df.groupby("Product")["Sales"].sum().sort_values()

# Create the bar chart
plt.figure(figsize=(10, 5))

plt.barh(product_sales.index, product_sales.values)

plt.title("Total Sales by Product", fontsize=16)
plt.xlabel("Total Sales (₹)")
plt.ylabel("Product")

plt.gca().xaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, pos: f"₹{x:,.0f}")
)

# Display sales values on the bars
for i, value in enumerate(product_sales.values):
    plt.text(value, i, f" ₹{value:,.0f}", va="center")

plt.tight_layout()
plt.show()
```

```python

# Calculate total sales for each region
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

# Create the chart
plt.figure(figsize=(9, 5))

plt.bar(region_sales.index, region_sales.values)

plt.title("Total Sales by Region", fontsize=16)
plt.xlabel("Region")
plt.ylabel("Total Sales (₹)")

plt.gca().yaxis.set_major_formatter(
    ticker.FuncFormatter(lambda x, pos: f"₹{x:,.0f}")
)

# Show sales values above each bar
for i, value in enumerate(region_sales.values):
    plt.text(i, value, f"₹{value:,.0f}", ha="center", va="bottom")

plt.tight_layout()
plt.show()

# Display the regional sales figures
print("Regional Sales Summary:")
print(region_sales)
```

```python

# CUSTOMER SALES AND PROFIT DASHBOARD

import matplotlib.pyplot as plt
import matplotlib.ticker as ticker

# Calculate key metrics
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order_ID"].nunique()
total_customers = df["Customer"].nunique()

# Prepare chart data
monthly_sales = (
    df.set_index("Order_Date")["Sales"]
    .resample("MS")
    .sum()
)
product_sales = df.groupby("Product")["Sales"].sum().sort_values()
region_sales = df.groupby("Region")["Sales"].sum().sort_values(
    ascending=False
)

# Create dashboard layout
fig, axes = plt.subplots(2, 2, figsize=(16, 10))
fig.suptitle(
    "CUSTOMER SALES AND PROFIT DASHBOARD",
    fontsize=19,
    fontweight="bold"
)

# Chart 1: Monthly sales
axes[0, 0].plot(
    monthly_sales.index,
    monthly_sales.values,
    marker="o",
    linewidth=2
)
axes[0, 0].set_title("Monthly Sales Trend")
axes[0, 0].set_ylabel("Sales (₹)")
axes[0, 0].set_xticks(monthly_sales.index)
axes[0, 0].set_xticklabels(
    monthly_sales.index.strftime("%b"),
    rotation=45
)
axes[0, 0].grid(alpha=0.3)

# Chart 2: Product sales
axes[0, 1].barh(product_sales.index, product_sales.values)
axes[0, 1].set_title("Sales by Product")
axes[0, 1].set_xlabel("Sales (₹)")

# Chart 3: Regional sales
axes[1, 0].bar(region_sales.index, region_sales.values)
axes[1, 0].set_title("Sales by Region")
axes[1, 0].set_ylabel("Sales (₹)")

# Chart 4: Profit by product
product_profit = df.groupby("Product")["Profit"].sum().sort_values()
axes[1, 1].barh(product_profit.index, product_profit.values)
axes[1, 1].set_title("Profit by Product")
axes[1, 1].set_xlabel("Profit (₹)")

# Format chart values as Indian rupees
for ax in axes.flat:
    ax.xaxis.set_major_formatter(
        ticker.FuncFormatter(lambda x, pos: f"₹{x:,.0f}")
    )
    ax.tick_params(axis="both", labelsize=9)

# Add summary metrics at the bottom
summary = (
    f"Total Sales: ₹{total_sales:,.0f}     |     "
    f"Total Profit: ₹{total_profit:,.0f}     |     "
    f"Orders: {total_orders}     |     "
    f"Customers: {total_customers}"
)

fig.text(0.5, 0.015, summary, ha="center", fontsize=11)

plt.tight_layout(rect=[0, 0.06, 1, 0.94])
plt.show()
```
